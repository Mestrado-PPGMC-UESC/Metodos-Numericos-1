# Implementação de Regressão Linear Múltipla

## 1. Definição do Problema
O objetivo desta seção é ajustar um modelo de regressão linear múltipla a um conjunto de dados $N$-dimensional. Conforme solicitado, utilizou-se a **Formulação Geral de Mínimos Quadrados** para derivar o sistema de equações lineares cujos coeficientes minimizam o erro quadrático total.

## 2. Metodologia: Equações Normais
Para encontrar os coeficientes ótimos $\mathbf{a} = [a_0, a_1, \dots, a_p]^T$, buscamos resolver o sistema linear decorrente da minimização dos resíduos. A abordagem matricial adotada baseia-se na equação:

$$
(\mathbf{X}^T \mathbf{X}) \mathbf{a} = \mathbf{X}^T \mathbf{y}
$$

Onde:
* $\mathbf{X}$ é a matriz de funções base, onde a primeira coluna é composta por $1$s (referente ao intercepto $a_0$) e as colunas subsequentes contêm os valores das variáveis independentes $x_i$.
* $\mathbf{y}$ é o vetor dos valores observados.
* $\mathbf{a}$ é o vetor de coeficientes que desejamos encontrar.

### Funções Base
O modelo assume uma combinação linear das variáveis de entrada. As funções base identificadas são:
* $\phi_0(\mathbf{x}) = 1$ (Intercepto)
* $\phi_i(\mathbf{x}) = x_i$ para $i = 1 \dots p$

## 3. Implementação Computacional

Para otimizar o processamento e evitar a alocação de memória de matrizes transpostas gigantes, o algoritmo foi implementado para construir diretamente os termos do sistema linear final:

1.  **Matriz A (Lado Esquerdo):** Corresponde a $\mathbf{X}^T \mathbf{X}$. Cada elemento $A_{ij}$ é calculado pelo produto escalar entre as colunas correspondentes das variáveis preditoras (ou vetor de 1s para o intercepto).
2.  **Vetor B (Lado Direito):** Corresponde a $\mathbf{X}^T \mathbf{y}$. Cada elemento $B_i$ é o produto escalar entre a coluna da variável $x_i$ e o vetor alvo $y$.
3.  **Resolução:** O sistema $\mathbf{A} \cdot \mathbf{a} = \mathbf{B}$ é resolvido utilizando álgebra linear numérica (`numpy.linalg.solve`).

### Estrutura de Dados
Os dados são gerenciados pela classe customizada `ListaPontosND`, que abstrai a complexidade de lidar com pontos em qualquer dimensão (2D, 3D, ..., ND), garantindo flexibilidade ao algoritmo.

## 4. Avaliação de Qualidade ($R^2$)
Para avaliar a qualidade do ajuste, foi implementado o cálculo do **Coeficiente de Determinação ($R^2$)**:

$$
R^2 = 1 - \frac{SQ_{res}}{SQ_{tot}} = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}
$$

Onde:
* $SQ_{res}$: Soma dos quadrados dos resíduos (variância não explicada pelo modelo).
* $SQ_{tot}$: Soma total dos quadrados (variância total dos dados).

```python
from utils import build_expression, calc_r_squared
from regressao_multipla import get_coeficientes
from pontoND import ListaPontosND
```

```python
coordenadas = [
    [1, 1, 6],
    [2, 1, 8],
    [1, 3, 10],
    [3, 4, 15]
]

lista_pontos = ListaPontosND.from_coordinates_list(coordenadas)
lista_pontos.print()
```

```python
coeficientes = get_coeficientes(lista_pontos)
expressao = build_expression('linear multipla', coeficientes)
r_squared = calc_r_squared(lista_pontos, expressao)

print("Coeficientes da regressão linear múltipla: ", coeficientes)
print(f"Expressão da regressão linear múltipla:\n  {expressao}")
print(f"R² da regressão linear múltipla: {r_squared}")
```
